```python
import os
import json
import re
import nltk
from konlpy.tag import Komoran

# Komoran 객체 생성
komoran = Komoran()

# NLTK의 영어 품사 태깅을 위한 데이터 다운로드
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

```

```python
# 1. JSON 파일 재귀적으로 탐색 및 전처리
def process_json_files(root_dir, output_file):
    label_map = {"O": 0}
    idx = 1

    with open(output_file, 'w', encoding='utf-8') as output_f:
        output_f.write("tokens,tags\n")

        for dirpath, _, filenames in os.walk(root_dir):
            for filename in filenames:
                if filename.endswith(".json"):
                    file_path = os.path.join(dirpath, filename)
                    try:
                        with open(file_path, 'r', encoding='utf-8') as f:
                            file_data = json.load(f)

                            # JSON 파일에서 dialogs 필드 확인
                            if "dialogs" not in file_data:
                                print(f"No 'dialogs' field in {file_path}")
                                continue

                            dialogs = file_data["dialogs"]

                            # dialogs 필드의 각 대화에 대해 전처리 수행
                            for dialog in dialogs:
                                if "text" not in dialog or "expression" not in dialog:
                                    continue

                                tokens, tags = tag_code_switching(dialog)

                                # 태그 업데이트
                                for tag in tags:
                                    if tag not in label_map:
                                        label_map[tag] = idx
                                        idx += 1

                                # 결과 저장
                                tokens_str = " ".join(tokens)
                                tags_str = " ".join(tags)
                                output_f.write(f"{tokens_str},{tags_str}\n")

                    except json.JSONDecodeError:
                        print(f"Error decoding JSON file: {file_path}")
                    except FileNotFoundError:
                        print(f"File not found: {file_path}")
                    except Exception as e:
                        print(f"Unexpected error for file {file_path}: {e}")

    return label_map

# 2. 코드스위칭 태깅 함수 (Komoran 사용)
def tag_code_switching(dialog):
    text = dialog["text"]
    expressions = {exp["form"]: exp["originalForm"] for exp in dialog.get("expression", [])}

    # Komoran 형태소 분석
    tokens = [token for token, pos in komoran.pos(text)]
    bio_tags = []

    for token in tokens:
        tag = "O"  # 기본 태그

        # 코드스위칭 여부 확인
        if token in expressions:
            original_form = expressions[token]
            # originalForm에서 영어 단어 추출
            english_parts = re.findall(r"[a-zA-Z]+", original_form)

            if english_parts:
                # 영어 코드스위칭 감지 및 품사 태깅
                pos_tag = nltk.pos_tag(english_parts)[0][1]
                tag = f"CS_EN_{pos_tag}"
            else:
                # 영어 부분이 없으면 한국어 품사 태깅
                pass
        # 영어 코드스위칭이 아니거나, 영어 부분이 없는 경우
        if tag == "O":  # 태그가 여전히 "O"인 경우에만 한국어 품사 태깅 수행
            pos_tag = komoran.pos(token)[0][1]  # Komoran 품사 태깅
            tag = f"KO_{pos_tag}"

        bio_tags.append(tag)

    return tokens, bio_tags

# 경로 설정 및 함수 호출
root_dir = r"/content/drive/MyDrive/한영혼합데이터/"
output_file = r"/content/drive/MyDrive/processed_data.csv"

# JSON 파일 전처리 및 저장
label_map = process_json_files(root_dir, output_file)
print(f"Data processing completed. Label map: {label_map}")
```

```python
!pip install transformers datasets

import pandas as pd
from datasets import Dataset
from transformers import BertTokenizer, BertForTokenClassification, Trainer, TrainingArguments

# CSV 파일 로드
csv_path = "/content/drive/MyDrive/processed_data.csv"
data = pd.read_csv(csv_path)

# tokens와 tags 열을 리스트 형식으로 변환
data['tokens'] = data['tokens'].apply(lambda x: x.split())
data['tags'] = data['tags'].apply(lambda x: x.split())

# 데이터셋 변환
dataset = Dataset.from_pandas(data)

# label_map 정의 (프로세싱 단계에서 반환된 label_map 사용)
label_map = {
    "O": 0,
    "KO_VV": 1,
    "KO_EF": 2,
    "CS_EN_NN": 3,
    "CS_EN_VB": 4,
    # 필요한 태그 추가
}

# 태그 인덱싱 함수 정의
def encode_tags(tags):
    return [label_map[tag] for tag in tags]

# 데이터셋에 태그 인덱싱 적용
dataset = dataset.map(lambda x: {'label_ids': encode_tags(x['tags'])})

# 토크나이저 및 모델 로드
model_name = "klue/bert-base"
tokenizer = BertTokenizer.from_pretrained(model_name)
model = BertForTokenClassification.from_pretrained(model_name, num_labels=len(label_map))

# 토큰화 함수 정
def tokenize_and_align_labels(examples):
    tokenized_inputs = tokenizer(examples['tokens'], truncation=True, is_split_into_words=True)

    labels = []
    for i, label in enumerate(examples['label_ids']):
        word_ids = tokenized_inputs.word_ids(batch_index=i)
        aligned_labels = []
        previous_word_idx = None
        for word_idx in word_ids:
            if word_idx is None:
                aligned_labels.append(-100)
            elif word_idx != previous_word_idx:
                aligned_labels.append(label[word_idx])
            else:
                aligned_labels.append(-100)
            previous_word_idx = word_idx
        labels.append(aligned_labels)

    tokenized_inputs['labels'] = labels
    return tokenized_inputs

# 데이터셋에 토큰화 및 라벨 정렬 적용
tokenized_dataset = dataset.map(tokenize_and_align_labels, batched=True)

# 학습 설정
training_args = TrainingArguments(
    output_dir="./results",
    evaluation_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=3,
    weight_decay=0.01,
    logging_dir="./logs",
    logging_steps=10,
)

# Trainer 생성
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset,
    eval_dataset=tokenized_dataset,
    tokenizer=tokenizer,
)

# 모델 학습
trainer.train()

# 모델 평가
eval_results = trainer.evaluate()
print("Evaluation Results:", eval_results)


```
