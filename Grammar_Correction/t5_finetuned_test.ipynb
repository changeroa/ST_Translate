```python
import torch
from transformers import AutoTokenizer, MarianMTModel

# 모델과 토크나이저 로드 (파인튜닝된 모델을 사용)
finetuned_model_name = "/content/drive/MyDrive/DAT프로젝트/finetuned-opus-mt-ko-en"  # 파인튜닝된 모델 경로
tokenizer = AutoTokenizer.from_pretrained(finetuned_model_name)
model = MarianMTModel.from_pretrained(finetuned_model_name)

def translate_text(text, tokenizer, model):
    # 텍스트를 토크나이즈
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True).to(model.device)

    # 모델로 번역
    with torch.no_grad():
        translated_tokens = model.generate(
            inputs["input_ids"],
            max_length=512,
            num_beams=6,
            early_stopping=True
        )

    # 번역된 텍스트 디토크나이즈
    translated_text = tokenizer.decode(translated_tokens[0], skip_special_tokens=True)

    return translated_text

# 번역할 텍스트
text = "오늘 아침에는 정말 바빴어. I had to finish some 보고서 for my class, but then I realized I forgot to bring my 노트북. 그래서 친구한테 연락했는데, she said, 'Don't worry, you can use mine."
#' 그래서 겨우 작업을 끝내고 학교에 갔는데, during the presentation, the 교수님 asked me a really tricky question in English, and I was like, 'Oh no, I need to think about this for a second.' 다행히 잘 대답해서 무사히 넘어갔어."

# 번역 실행
translated = translate_text(text, tokenizer, model)
print(f"Original: {text}")
print(f"Translated: {translated}")
```
