```python
from transformers import pipeline, AutoTokenizer
from datasets import Dataset
import torch
import json

# 데이터 파일 경로
input_file_path = "/content/drive/MyDrive/DAT프로젝트/kor_en_panyo_corpus.jsonl"
output_file_path = "/content/drive/MyDrive/DAT프로젝트/processed_korean_pangyo_pipeline.json"

# GPU 설정
device = 0 if torch.cuda.is_available() else -1
print(f"Using device: {'GPU' if device == 0 else 'CPU'}")

# Translator pipeline 및 tokenizer 초기화
translator = pipeline("translation", model="Helsinki-NLP/opus-mt-ko-en", device=device, max_length=400)
tokenizer = AutoTokenizer.from_pretrained("Helsinki-NLP/opus-mt-ko-en")

# 긴 문장 나누기 함수
def split_into_token_chunks(text, tokenizer, max_length=400):
    tokens = tokenizer(text, return_tensors="pt", truncation=False)["input_ids"].squeeze(0)
    chunks = []
    for i in range(0, len(tokens), max_length - 50):  # 안전 마진 추가
        chunk = tokens[i:i + max_length - 50]
        chunks.append(tokenizer.decode(chunk, skip_special_tokens=True))
    return chunks

# JSONL 파일 읽기
data = []
with open(input_file_path, 'r', encoding='utf-8') as file:
    for line in file:
        data.append(json.loads(line))

# Dataset 생성
dataset = Dataset.from_list(data)

# 번역 함수
def translate_entry(entry):
    korean_text = entry["korean"]
    pangyo_text = entry["pangyo"]

    # 긴 문장 나누기
    chunks = split_into_token_chunks(korean_text, tokenizer, max_length=400)

    # 번역 수행
    translated_chunks = translator(chunks)
    translated_text = " ".join([result["translation_text"] for result in translated_chunks])

    return {"source": pangyo_text, "target": translated_text}

# Dataset에 map 적용
processed_dataset = dataset.map(translate_entry)

# 결과 저장
processed_data = processed_dataset.to_list()
with open(output_file_path, "w", encoding="utf-8") as f:
    json.dump(processed_data, f, ensure_ascii=False, indent=4)

print(f"Processed file saved to {output_file_path}")
```

```python
from datasets import Dataset, DatasetDict
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, Seq2SeqTrainingArguments, Seq2SeqTrainer
from transformers import DataCollatorForSeq2Seq
import os
os.environ["WANDB_DISABLED"] = "true"

# GPU 확인
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# 모델 및 토크나이저 로드
model_name = "Helsinki-NLP/opus-mt-ko-en"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)  # CPU에 로드
#model.gradient_checkpointing_enable()
model = model.to(device)   # 모델을 GPU로 이동

# 데이터셋 로드
def preprocess_function(examples):
    inputs = examples["source"]
    targets = examples["target"]
    model_inputs = tokenizer(inputs, max_length=512, truncation=True)
    with tokenizer.as_target_tokenizer():
        labels = tokenizer(targets, max_length=512, truncation=True)
    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

# Train, eval 데이터셋 생성
train_dataset = Dataset.from_list(train_data)
eval_dataset = Dataset.from_list(eval_data)

# Preprocess 데이터셋
tokenized_train = train_dataset.map(preprocess_function, batched=True)
tokenized_eval = eval_dataset.map(preprocess_function, batched=True)

# DataCollator 정의
data_collator = DataCollatorForSeq2Seq(tokenizer, model=model)

# Training Arguments (GPU 활성화 포함)
training_args = Seq2SeqTrainingArguments(
    output_dir="./results",
    eval_strategy="epoch",
    learning_rate=5e-5,
    per_device_train_batch_size=4,  # GPU 메모리에 맞게 조정
    per_device_eval_batch_size=4,
    num_train_epochs=3,
    save_strategy="epoch",
    save_total_limit=2,
    predict_with_generate=True,
    logging_dir="./logs",
    logging_steps=50,
    load_best_model_at_end=True,
    metric_for_best_model="eval_loss",
    fp16=True,  # Mixed precision for faster training on GPU
    half_precision_backend="auto",  # AMP 설정
    run_name="my_experiment"
)

# Trainer 정의
trainer = Seq2SeqTrainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_eval,
    tokenizer=tokenizer,
    data_collator=data_collator,
)

# 모델 훈련
trainer.train()

# 모델 저장
model.save_pretrained("/content/drive/MyDrive/DAT프로젝트/finetuned-opus-mt-ko-en")
tokenizer.save_pretrained("/content/drive/MyDrive/DAT프로젝트/finetuned-opus-mt-ko-en")
```

```python
import torch
from transformers import AutoTokenizer, MarianMTModel

# 모델과 토크나이저 로드 (파인튜닝된 모델을 사용)
finetuned_model_name = "./finetuned-opus-mt-ko-en"  # 파인튜닝된 모델 경로
tokenizer = AutoTokenizer.from_pretrained(finetuned_model_name)
model = MarianMTModel.from_pretrained(finetuned_model_name)

def translate_text(text, tokenizer, model):
    # 텍스트를 토크나이즈
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True).to(model.device)

    # 모델로 번역
    with torch.no_grad():
        translated_tokens = model.generate(
            inputs["input_ids"],
            max_length=512,
            num_beams=6,
            early_stopping=True
        )

    # 번역된 텍스트 디토크나이즈
    translated_text = tokenizer.decode(translated_tokens[0], skip_special_tokens=True)

    return translated_text

# 번역할 텍스트
text = "For me.가 Personal적인 Experience을 가지거나 Prefer를 Organisation할 Number는 없는 Man-in-law Intelligence Model이M.로, "

# 번역 실행
translated = translate_text(text, tokenizer, model)
print(f"Original: {text}")
print(f"Translated: {translated}")
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# 훈련된 모델 및 토크나이저 로드
model_path = "./finetuned-opus-mt-ko-en"
tokenizer = AutoTokenizer.from_pretrained(model_path)
model = AutoModelForSeq2SeqLM.from_pretrained(model_path).to("cuda")  # GPU 사용

# 번역 함수
def translate_text(text, tokenizer, model):
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True).to("cuda")
    outputs = model.generate(inputs["input_ids"], max_length=512, num_beams=6, early_stopping=True)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

# 테스트 문장
test_sentence = "개양귀비는 양귀비과에 속하는 식물로, 보통 30~80 cm까지 grow합니다. 대부분은 붉은 또는 주황색의 꽃을 bloom하지만, white 꽃도 있습니다. 유럽에서는 long time 동안 농작물로 cultivate해 왔으며, 풍성한 harvest를 상징하는 꽃으로 여겨 왔습니다."

translated_text = translate_text(test_sentence, tokenizer, model)
print(f"Original: {test_sentence}")
print(f"Translated: {translated_text}")
```
