```python
import json
from keybert import KeyBERT
from sklearn.feature_extraction import text
from datasets import load_dataset

# 데이터 로드
ds = load_dataset("wikimedia/wikipedia", "20231101.en", split='train')

# KeyBERT 모델 로드
kw_model = KeyBERT()

# 사용자 정의 불용어 목록
additional_stop_words = [
    'et', 'al', 'figure', 'table', 'data', 'results', 'conclusion',
    "examine", "analyze", "assess", "investigate", "describe",
    "identify", "evaluate", "discuss", "present", "propose", "consider",
    "illustrate", "explain", "conclude", "provide", "show", "suggest",
    "demonstrate", "indicate", "include", "category", "references",
    "cite", "page", "article", "source", "section"
]
stop_words = text.ENGLISH_STOP_WORDS.union(additional_stop_words)

def extract_keywords_from_text(text_content, top_n=10):
    """
    텍스트에서 상위 N개의 키워드를 추출합니다.
    """
    # KeyBERT를 사용해 키워드 추출
    keywords = kw_model.extract_keywords(text_content, top_n=top_n, stop_words=stop_words)
    return keywords

# 배치 처리로 키워드 추출 함수
def batch_process_documents(dataset, batch_size=10, top_n=10):
    """
    DatasetDict 객체의 모든 문서에 대해 배치로 키워드를 추출합니다.
    """
    total_documents = len(dataset)
    print(f"총 문서 수: {total_documents}")

    # 배치별로 문서를 처리
    for i in range(0, total_documents, batch_size):
        batch_documents = dataset.select(range(i, min(i + batch_size, total_documents)))

        print(f"\n--- Batch {i // batch_size + 1} ---")

        for document in batch_documents:
            document_id = document.get('id', 'Unknown ID')
            title = document.get('title', 'No Title')
            text_content = document.get('text', '')

            if text_content:
                keywords = extract_keywords_from_text(text_content, top_n=top_n)
                print(f"\n문서 ID: {document_id}")
                print(f"제목: {title}")
                print("추출된 키워드:")
                for word, score in keywords:
                    print(f"{word}: {score:.4f}")
            else:
                print(f"문서 ID: {document_id} - 본문 텍스트가 없습니다.")

# 데이터셋에서 배치 처리로 키워드 추출
batch_process_documents(ds, batch_size=10, top_n=10)
```
